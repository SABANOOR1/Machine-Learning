# **Dataset Exploration**

```python
# Task 1: Load the dataset and briefly inspect it (shape, dtypes, first few rows, summary statistics). 
import pandas as pd
df=pd.read_csv("house-prices-advanced-regression-techniques/train.csv")
df.head(10)
df.dtypes
df.shape
df.info()
df.describe()
```

# **Feature and Target Identification**

```python
# Task 2: Identify the dependent variable and list the independent variables, split explicitly into numerical features and categorical features. 
y=df["SalePrice"]
x=df.drop(columns=["SalePrice","Id"])

numericalFeatures=x.select_dtypes(include=["int64","float64"]).columns.tolist()
categoricalFeatures=x.select_dtypes(include=["str"]).columns.tolist()

print("Dependent Features are:")
print("SalePrice")
print("--------------------------")

print("Numerical Features are:")
print(numericalFeatures)
print("--------------------------")
print("Categorical Features are:")
print(categoricalFeatures)
```

# **Problem Definition**

# Task 3: In 1–2 sentences, state why this is a regression problem rather than a classification problem, and what that changes about how you'll evaluate the model later. 
This is a regression problem because the dependent variable, SalePrice, is a continuous numerical value that represents the price of a house. Therefore, the model will be evaluated using regression metrics such as MAE, MSE, RMSE, or R² rather than classification metrics such as accuracy or precision.

# **Data Preparation**

```python
# Task 4: Prepare X (features) and y (price) for modeling: encode every categorical column into numeric form, and briefly justify your choice of encoding method for each type of column. 
X_encoded=pd.get_dummies(x,columns=categoricalFeatures, drop_first=True)
X_encoded=X_encoded.fillna(X_encoded.median(numeric_only=True))
X_encoded=X_encoded.astype(int)

print("Original Shape:", x.shape)
print("Encoded Shape:", X_encoded.shape)
```

Justification:
One-hot encoding is used for categorical features because the categories do not necessarily have a meaningful numerical order. Each category is represented by a separate binary column, allowing the regression model to use the information without introducing a false ranking between categories.

```python
# Task 5: Confirm X is fully numeric with no missing values. 
print("All columns numeric:", X_encoded.dtypes.apply(lambda dtype: dtype != "object").all())
print("Total missing values:",X_encoded.isnull().sum().sum())
```

# **Train/Test Split**

```python
# Task 6: Split X and y into training and test sets (e.g. an 80/20 split) and report the resulting shapes. 
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test=train_test_split(X_encoded, y, test_size=0.20, random_state=42)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", Y_train.shape)
print("y_test shape:", Y_test.shape)
```

# **Linear Regression Model**

```python
# Task 7: Train a LinearRegression model on the training data. 
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train,Y_train)
print("Model trained successfully")
```

# **Predictions**

```python
# Task 8: Generate predicted prices on the test set, and display a small table comparing actual vs. predicted price for 10 test houses.
Y_predicted=model.predict(X_test)

comparison=pd.DataFrame({
    "Actual Price": Y_test.values,
    "Predicted Price": Y_predicted
})

print(comparison.head(10))
    
```

# **Coefficients and Intercept**

```python
# Task 9: Extract the intercept and all coefficients. Build a table: Feature | Coefficient | Interpretation, with a one-line, plain-English interpretation for each feature (e.g. "each additional bathroom is associated with a price increase of approximately ___, holding other features constant"). 
intercept=model.intercept_
coefficient=model.coef_

print("Intercept:", intercept)
print("coefficient:", len(coefficient))

coefficient_table=pd.DataFrame({
    "Feature":X_train.columns,
    "Coefficient": coefficient
})

print(coefficient_table)
```

```python
coefficient_table["Absolute Coefficient"] = coefficient_table["Coefficient"].abs()

top_features = coefficient_table.sort_values(
    by="Absolute Coefficient",
    ascending=False
).head(3)

print(top_features[["Feature", "Coefficient"]])
```

# Task 10: Does the intercept have a sensible real-world interpretation for this dataset? Why or why not?

The intercept has limited real-world meaning because it represents the predicted house price when all independent variables are zero. A house with zero values for all of these characteristics is not a realistic observation, so the intercept mainly serves as a mathematical baseline for the regression model rather than a meaningful house price.

# **Coefficients and Intercept**

```python
# Task 11: Compute residuals on the test set and plot them (e.g. residuals vs. predicted values, and a histogram of residuals). Do the residuals look randomly scattered, or is there a visible pattern? 
import matplotlib.pyplot as plt
residuals=Y_test-Y_predicted
print("first 10 residuals:")
print(residuals.head(10))

plt.figure(figsize=(8,5))
plt.scatter(Y_predicted,residuals)
plt.xlabel("Predicted Price")
plt.ylabel("Residuals")
plt.title("Residuals vs Predicted Prices")
plt.show()

plt.figure(figsize=(8,5))
plt.hist(residuals, bins=30)
plt.xlabel("Residual")
plt.ylabel("Frequency")
plt.title("Distribution of Residuals")
plt.show()
```

# **Regression Metrics**

```python
# Task 12: Compute MAE, MSE, RMSE, and R² on the test set. Report them in a small table. 
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae=mean_absolute_error(Y_test,Y_predicted)
mse=mean_squared_error(Y_test,Y_predicted)
rmse=np.sqrt(mse)
r2=r2_score(Y_test,Y_predicted)

metrics_table=pd.DataFrame({
    "Metric":["MAE","MSE","RMSE","R2"],
    "Value":[mae,mse,rmse,r2]
})

print(metrics_table)
```

# Task 13: In your own words, what does each metric tell you, and which one would you show to a non-technical stakeholder? Why? 

MAE: Shows the average absolute difference between the actual and predicted house prices. It is easy to understand because it is in the same unit as the price.

MSE: Calculates the average squared prediction error. Because errors are squared, large errors have a much greater effect on this metric.

RMSE: The square root of MSE. It is also in the same unit as house prices, but it gives more importance to large errors.

R²: Shows how much of the variation in house prices is explained by the model. A value closer to 1 generally indicates that the model explains more of the variation.



# **Training vs Testing Performance**

```python
# Task 14: Compute MAE, MSE, RMSE, and R² on the training set, and place them side by side with your Part 5 test-set metrics in one comparison table.
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

Y_train_predict=model.predict(X_train)
train_mae = mean_absolute_error(Y_train, Y_train_predict)
train_mse = mean_squared_error(Y_train, Y_train_predict)
train_rmse = np.sqrt(train_mse)
train_r2 = r2_score(Y_train, Y_train_predict)

comparison_table = pd.DataFrame({
    "Metric": ["MAE", "MSE", "RMSE", "R²"],
    "Training Set": [train_mae, train_mse, train_rmse, train_r2],
    "Testing Set": [mae, mse, rmse, r2]
})
print(comparison_table)
```

# Task 15: Is there a large gap between training and testing performance? What would a large gap suggest about overfitting or generalization, and what does your actual result suggest here? 

Yes, there is a noticeable difference between the training and testing results. The training R² is 0.9362, while the testing R² is 0.6837. The testing RMSE is also much higher than the training RMSE. This difference suggests that the model performs better on the data it was trained on than on new, unseen data, which can be a sign of overfitting. However, the testing R² of 0.6837 shows that the model is still able to explain a good amount of the variation in house prices.
