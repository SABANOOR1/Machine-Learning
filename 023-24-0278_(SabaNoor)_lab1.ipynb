```python
# Exercise Task 2
def is_prime(n):
    if n==2: 
        return True

    if n%2==0:
        return False
    
    for i in range(n//2,1,-1):
       if n%i==0:
          return False
       
    return True

e= is_prime(4)
print(e)
```

```python
# Exercise Task 4
def remove_duplicates(numbers):
    Result=[]

    for i in numbers:
        if i not in Result:
            Result.append(i)

    return Result

a=[1,2,3,4,5,4,3]
print(remove_duplicates(a))
```

```python
# Exercise Task 5
def Product(*args):
    result=1
    for i in args:
        result=result*i
    return result

print(Product(8,6,5,4))

```

```python
# Exercise Task 23
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "name": ["Yahya", "Saba", "Noor", "Hamna", "Ahmed", "Sara", "Usman", "Ayesha"],
    "section": ["A", "A", "A", "B", "A", "B", "A", "B"],
    "marks": [85, 72, 91, 60, 78, 88, 95, 74]
}
df=pd.DataFrame(data)
result=df.groupby("section")["marks"].agg(["mean","max"])
result.plot(kind="bar")
plt.xlabel("section")
plt.ylabel("marks")
plt.title("Mean and maximum According to the Section")
plt.show()
```

```python
# Exercise Task 24
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Number of Students": [4,6,8,9,3,4,5,6],
    "marks": [85, 72, 91, 68, 78, 88, 95, 74]
}
df=pd.DataFrame(data)
df["marks"].plot(kind="hist",bins=8)
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks")
plt.show()
```

```python
# Exercise Task 25
import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot(df["marks"])
axes[0].set_title("Marks Trend")
axes[0].set_xlabel("Student")
axes[0].set_ylabel("Marks")

axes[1].scatter(df["marks"], range(len(df)))
axes[1].set_title("Marks vs Student")
axes[1].set_xlabel("Marks")
axes[1].set_ylabel("Student Index")

plt.show()
```
